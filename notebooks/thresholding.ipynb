```python
import cv2
import numpy as np
```

```python
orig_gray = cv2.imread('../angio-6/slice.png', cv2.IMREAD_GRAYSCALE)
cv2.imshow('Original gray image', orig_gray)
cv2.waitKey(0)
```

```python
orig_color = cv2.imread('../angio-6/slice.png', cv2.IMREAD_COLOR)
cv2.imshow('Original color image', orig_color)
cv2.waitKey(0)
```

```python
def thresholding(orig, name='Tresholded image'):
    img =  orig.copy()
    #img = img / 255.0
    _, tresh = cv2.threshold(img, 127, 255,cv2.THRESH_BINARY)
    cv2.imshow(name, tresh)
    cv2.waitKey(0)

thresholding(orig_gray)
```

```python
def adaptive_tresholding(orig, name= 'Adaptive thresholded image'):
    img = orig.copy()
    thresh = cv2.adaptiveThreshold(img,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY,15,3)
    cv2.imshow(name, thresh)
    cv2.waitKey(0)

adaptive_tresholding(orig_gray)
```

```python
def blur(orig, name='Gaussian blurred image'):
    img = orig.copy()
    blurred = cv2.GaussianBlur(img,(7,7), 5, 5)
    return blurred
    cv2.imshow(name, blurred)
    cv2.waitKey(0)


adaptive_tresholding(blur(orig_gray))
```

```python
def green(orig, name='Green channel image'):
    img = orig.copy()
    img[:,:,0] = 0
    img[:,:,2] = 0
    return img
    cv2.imshow(name, img)
    cv2.waitKey(0)

green(orig_color)     
```

```python
def CLAHE(orig, name='Contrast Limited AHE'):
    green_img = green(orig)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(15,15))
    lab = cv2.cvtColor(green_img,cv2.COLOR_BGR2Lab)
    lab_planes = list(cv2.split(lab))
    lab_planes[0] = clahe.apply(lab_planes[0])
    lab = cv2.merge(lab_planes)
    green_clahe = cv2.cvtColor(lab,cv2.COLOR_Lab2BGR)
    return green_clahe
    cv2.imshow(name, green_clahe)
    cv2.waitKey(0)

CLAHE(orig_color)
```

```python
green_clahe_img = CLAHE(orig_color)
gray_green_clahe_img = cv2.cvtColor(green_clahe_img,cv2.COLOR_BGR2GRAY)
adaptive_tresholding(blur(gray_green_clahe_img))
```
