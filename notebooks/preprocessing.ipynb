```python
import numpy as np
import cv2
```

```python
PATH2IMG = 'C://Users/abram/Downloads/angio-6/slice.png'
```

```python
img = cv2.imread(PATH2IMG, cv2.IMREAD_GRAYSCALE)
```

```python
def homomorphic_filter(src, d0=12, r1=0.5, rh=2, c=4, h=2.0, l=0.5):
    gray = src.copy()
    if len(src.shape) > 2:
        gray = cv2.cvtColor(src, cv2.COLOR_BGR2GRAY)
    gray = np.float64(gray) 
    rows, cols = gray.shape
    gray_fft = np.fft.fft2(gray)
    gray_fftshift = np.fft.fftshift(gray_fft)
    dst_fftshift = np.zeros_like(gray_fftshift)
    M, N = np.meshgrid(np.arange(-cols // 2, cols // 2), np.arange(-rows//2, rows//2))
    D = np.sqrt(M ** 2 + N ** 2)
    Z = (rh - r1) * (1 - np.exp(-c * (D ** 2 / d0 ** 2))) + r1
    dst_fftshift = Z * gray_fftshift
    dst_fftshift = (h - l) * dst_fftshift + l
    dst_ifftshift = np.fft.ifftshift(dst_fftshift)
    dst_ifft = np.fft.ifft2(dst_ifftshift)
    dst = np.real(dst_ifft)
    dst = np.uint8(np.clip(dst, 0, 255))

    return dst

img2draw = img.copy()
hommorphic = homomorphic_filter(img2draw)
cv2.imshow('hommorphic',hommorphic)
cv2.waitKey(0)
```

```python
def normalize_cv(src):
    dst = np.zeros((512,512))
    dst = cv2.normalize(src,dst,128,255,cv2.NORM_MINMAX)
    dst = cv2.GaussianBlur(dst,(7,7),1.4)
    return dst


img2draw = img.copy()
norm = normalize_cv(img2draw)
cv2.imshow('normalize', norm)
cv2.waitKey(0)
```

```python
edges = cv2.Canny(norm,128,255)
cv2.imshow('128 255', edges)

edges = cv2.Canny(norm,50,100)
cv2.imshow('50 100', edges)

edges = cv2.Canny(norm,10,60)
cv2.imshow('10 50', edges)

cv2.waitKey(0)
```
