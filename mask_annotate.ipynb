```python
from pycocotools.coco import COCO
from PIL import Image
import matplotlib.pyplot as plt
from torchvision.transforms import ToTensor 
import os

coco = COCO(path)

img_dir = 'C:/Users/abram/DataspellProjects/VCD/arcade_challenge_datasets/dataset_phase_1/stenosis_dataset/sten_train/images'
img = coco.imgs[2]

pil_img = np.array(Image.open(os.path.join(img_dir, img['file_name'])).convert('L'))
cv_img = cv2.imread(os.path.join(img_dir, img['file_name']),cv2.IMREAD_GRAYSCALE)

print(ToTensor()(pil_img).shape)
print(ToTensor()(cv_img).shape)
```
